```python
import pandas as pd
import numpy as np

df = pd.read_excel("../Tick_Sightings.xlsx")
```

```python
CITY_COORDS = {
    "London":      (51.509865, -0.118092),
    "Birmingham":  (52.489471, -1.898575),
    "Glasgow":     (55.860916, -4.251433),
    "Manchester":  (53.483959, -2.244644),
    "Liverpool":   (53.400002, -2.983333),
    "Leeds":       (53.801277, -1.548567),
    "Sheffield":   (53.383331, -1.466667),
    "Edinburgh":   (55.953251, -3.188267),
    "Bristol":     (51.454514, -2.587910),
    "Cardiff":     (51.481583, -3.179090),
    "Nottingham":  (52.95, -1.15),
    "Leicester":   (52.633331, -1.133333),
    "Southampton": (50.909698, -1.404351),
    "Newcastle":   (54.966667, -1.6),
}
```

```python
def add_coords_with_jitter(df, jitter=0.05):
    df = df.copy()
    lats, lons = [], []
    for loc in df["location"]:
        base_lat, base_lon = CITY_COORDS.get(loc, (54.0, -2.0))
        lats.append(base_lat + np.random.uniform(-jitter, jitter))
        lons.append(base_lon + np.random.uniform(-jitter, jitter))
    df["latitude"] = lats
    df["longitude"] = lons
    return df
```

```python
synthetic = df.drop(columns=["id"], errors="ignore")
synthetic = synthetic.sample(5000, replace=True).reset_index(drop=True)
synthetic["id"] = range(1, len(synthetic) + 1)
synthetic = add_coords_with_jitter(synthetic)
synthetic.to_csv("../synthetic_sightings.csv", index=False)
```
